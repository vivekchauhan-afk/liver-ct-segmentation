```python
from pathlib import Path
from collections import defaultdict
import cv2
import numpy as np


# =========================================================
# DATASET PATH
# =========================================================

IMAGE_DIR = Path(
    r"C:\Users\vivek chauhan\Desktop\dsa babbar\Videos\archive (26)\last_dataset\images"
)

MASK_DIR = Path(
    r"C:\Users\vivek chauhan\Desktop\dsa babbar\Videos\archive (26)\last_dataset\masks"
)


# =========================================================
# GET IMAGE FILES
# =========================================================

image_paths = sorted(IMAGE_DIR.glob("*.png"))

print("=" * 60)
print("VOLUME-LEVEL MASK ANALYSIS")
print("=" * 60)

print("Total images:", len(image_paths))


# =========================================================
# ANALYZE MASKS BY VOLUME
# =========================================================

volume_stats = defaultdict(
    lambda: {
        "total": 0,
        "empty": 0,
        "non_empty": 0
    }
)


for i, image_path in enumerate(image_paths):

    # Example:
    # Volume-000-000.png
    # Volume-000-001.png

    parts = image_path.stem.split("-")

    if len(parts) < 3:
        continue

    volume_id = f"{parts[0]}-{parts[1]}"

    mask_path = MASK_DIR / image_path.name

    mask = cv2.imread(
        str(mask_path),
        cv2.IMREAD_GRAYSCALE
    )

    if mask is None:
        continue

    # Same resizing used in preprocessing
    mask = cv2.resize(
        mask,
        (128, 128),
        interpolation=cv2.INTER_NEAREST
    )

    volume_stats[volume_id]["total"] += 1

    if np.sum(mask > 127) == 0:
        volume_stats[volume_id]["empty"] += 1
    else:
        volume_stats[volume_id]["non_empty"] += 1


# =========================================================
# VOLUME INFORMATION
# =========================================================

print("\n" + "=" * 60)
print("VOLUME INFORMATION")
print("=" * 60)

print(
    "Number of volumes analyzed:",
    len(volume_stats)
)


# =========================================================
# FIRST 10 VOLUMES
# =========================================================

print("\nFirst 10 volumes:")

for volume_id in list(volume_stats.keys())[:10]:

    stats = volume_stats[volume_id]

    total = stats["total"]
    empty = stats["empty"]
    non_empty = stats["non_empty"]

    percentage = (non_empty / total) * 100

    print(
        f"{volume_id}: "
        f"total={total}, "
        f"empty={empty}, "
        f"non-empty={non_empty}, "
        f"non-empty%={percentage:.2f}%"
    )


# =========================================================
# OVERALL VOLUME STATISTICS
# =========================================================

non_empty_percentages = []

for volume_id, stats in volume_stats.items():

    total = stats["total"]
    non_empty = stats["non_empty"]

    percentage = (non_empty / total) * 100

    non_empty_percentages.append(percentage)


print("\n" + "=" * 60)
print("OVERALL VOLUME STATISTICS")
print("=" * 60)

print(
    "Minimum non-empty percentage:",
    round(min(non_empty_percentages), 2)
)

print(
    "Maximum non-empty percentage:",
    round(max(non_empty_percentages), 2)
)

print(
    "Average non-empty percentage:",
    round(np.mean(non_empty_percentages), 2)
)

print(
    "Median non-empty percentage:",
    round(np.median(non_empty_percentages), 2)
)


# =========================================================
# VOLUMES WITH VERY FEW LIVER SLICES
# =========================================================

print("\n" + "=" * 60)
print("VOLUMES WITH LOW LIVER PRESENCE")
print("=" * 60)

low_presence_found = False

for volume_id, stats in volume_stats.items():

    total = stats["total"]
    non_empty = stats["non_empty"]

    percentage = (non_empty / total) * 100

    if percentage < 10:

        low_presence_found = True

        print(
            f"{volume_id}: "
            f"{non_empty}/{total} "
            f"({percentage:.2f}%) non-empty"
        )


if not low_presence_found:
    print("No volume has less than 10% non-empty slices.")


# =========================================================
# FINAL
# =========================================================

print("\n" + "=" * 60)
print("VOLUME-LEVEL MASK ANALYSIS COMPLETED")
print("=" * 60)
```
