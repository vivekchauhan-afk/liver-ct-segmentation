```python
from pathlib import Path
from collections import Counter


# =========================================================
# DATASET PATHS
# =========================================================

IMAGE_DIR = Path(
    r"C:\\Users\\vivek chauhan\\Desktop\\dsa babbar\\Videos\\archive (26)\\last_dataset\\images"
)

MASK_DIR = Path(
    r"C:\\Users\\vivek chauhan\\Desktop\\dsa babbar\\Videos\\archive (26)\\last_dataset\\masks"
)


# =========================================================
# CHECK DATASET
# =========================================================

print("=" * 60)
print("VOLUME-LEVEL DATASET ANALYSIS")
print("=" * 60)

print("Image folder:", IMAGE_DIR)
print("Image folder exists:", IMAGE_DIR.exists())

print("Mask folder :", MASK_DIR)
print("Mask folder exists :", MASK_DIR.exists())


if not IMAGE_DIR.exists():
    raise ValueError("Image folder does not exist.")

if not MASK_DIR.exists():
    raise ValueError("Mask folder does not exist.")


# =========================================================
# GET IMAGES
# =========================================================

image_paths = sorted(IMAGE_DIR.glob("*.png"))

print("\nTotal images:", len(image_paths))


# =========================================================
# EXTRACT VOLUME IDs
# =========================================================

volume_ids = []

for path in image_paths:

    # Example:
    # Volume-000-000.png
    # Volume-000-001.png

    parts = path.stem.split("-")

    if len(parts) >= 3 and parts[0].lower() == "volume":

        volume_id = f"{parts[0]}-{parts[1]}"

        volume_ids.append(volume_id)


# =========================================================
# VOLUME INFORMATION
# =========================================================

print("\n" + "=" * 60)
print("VOLUME INFORMATION")
print("=" * 60)

print("Images with recognizable volume IDs:", len(volume_ids))
print("Unique volumes:", len(set(volume_ids)))


# =========================================================
# COUNT SLICES PER VOLUME
# =========================================================

volume_counts = Counter(volume_ids)

print("\n" + "=" * 60)
print("SLICES PER VOLUME")
print("=" * 60)

if volume_counts:

    counts = list(volume_counts.values())

    print("Minimum slices per volume:", min(counts))
    print("Maximum slices per volume:", max(counts))
    print("Average slices per volume:", sum(counts) / len(counts))

    sorted_counts = sorted(counts)

    median = sorted_counts[len(sorted_counts) // 2]

    print("Median slices per volume:", median)

    print("\nFirst 10 volumes:")

    for volume, count in list(volume_counts.items())[:10]:

        print(volume, "->", count, "slices")


# =========================================================
# SAMPLE FILENAMES
# =========================================================

print("\n" + "=" * 60)
print("SAMPLE FILENAMES")
print("=" * 60)

for path in image_paths[:10]:

    print(path.name)


# =========================================================
# VOLUME ID RANGE
# =========================================================

unique_volumes = sorted(set(volume_ids))

print("\n" + "=" * 60)
print("VOLUME ID RANGE")
print("=" * 60)

if unique_volumes:

    print("First volume:", unique_volumes[0])

    print("Last volume :", unique_volumes[-1])

    print("\nFirst 20 volume IDs:")

    for volume in unique_volumes[:20]:

        print(volume)


# =========================================================
# FINAL SUMMARY
# =========================================================

print("\n" + "=" * 60)
print("FINAL SUMMARY")
print("=" * 60)

print("Total image slices:", len(image_paths))

print(
    "Recognized volume slices:",
    len(volume_ids)
)

print(
    "Unique volumes:",
    len(set(volume_ids))
)

if volume_counts:

    counts = list(volume_counts.values())

    print(
        "Minimum slices/volume:",
        min(counts)
    )

    print(
        "Maximum slices/volume:",
        max(counts)
    )

    print(
        "Average slices/volume:",
        round(sum(counts) / len(counts), 2)
    )

print("\nVolume-level analysis completed.")
```
